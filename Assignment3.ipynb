```python
# Import pandas and numpy with their aliases
import pandas as pd
import numpy as np
# Create a variable a = pd.Series([ 100, 200, 300, 400])
a = pd.Series([100, 200, 300, 400])
# Print a, and data type
print(a)
print(type(a))
# Using indexing access the element 300 from the series a.
print(a[2])
# What are the values of index for series a?
print(a.index)
# Change the index to  ['c', 'a', 'b', 'd']
a.index = ['c', 'a', 'b', 'd']
print(a)
# Access the value in the series with index 'd'
print(a['d'])
# Sort the values wrt to the index and print it
print(a.sort_index())
# Create a new Pandas Series b having index as 'e', 'f', and 'g' and value 800,450,100 and print it
b = pd.Series([800, 450, 100], index=['e', 'f', 'g'])
print(b)
# Append b series at the end of a series
a = a.append(b)
# print a again after appending b into it
print(a)
# Sort the values in descending order of a and print the index of the sorted series
a = a.sort_values(ascending=False)
print(a.index)
# Create a pandas dataframe df from the series 'a' that we used in the last section, print the dataframe
df = pd.DataFrame(a)
print(df)
# What is the shape of the datafarme
print(df.shape)
# What is the index of the dataframe, is it same as the series 'a'
print(df.index)
# print the head and tail of the dataframe.
print(df.head())
print(df.tail())
# Rename the column of the dataframe as 'points'
df.columns = ['points']
print(df)
# Create fruits array
fruits_array = np.random.choice(['orange', 'mango', 'apple'], size=7)
print(fruits_array)
# Create series fruits out of fruits array
fruits = pd.Series(fruits_array)
print(fruits)
# Change the index of fruits to the index of dataframe df
fruits.index = df.index
print(fruits)
# Add this fruits series as a new column to the dataframe df with its column name as 'fruits'
df['fruits'] = fruits
print(df.head())
# Create a dataframe  d1 where the cols are  ‘city’ : [‘Chandigarh’, ‘Delhi’, ‘Kanpur’, ‘Chennai’, ‘Manali’ ] and ‘Temperature’ : [15, 22, 20, 26,-2]
d1 = pd.DataFrame({
    'city': ['Chandigarh', 'Delhi', 'Kanpur', 'Chennai', 'Manali'],
    'Temperature': [15, 22, 20, 26, -2]
})
# Print(d1)
print(d1)
# What is the shape of d1.
print(d1.shape)
# Set city = d1['city']
city = d1['city']
# print city
print(city)
# What is the type of city.
print(type(city))
# Create another datafeame 'd2'
d2 = pd.DataFrame({
    'city': ['Bengalaru', 'Coimbatore', 'Srirangam', 'Pondicherry'],
    'Temperature': [24, 35, 36, 39]
})
# print the shape of this dataframe
print(d2.shape)
# merge the two dataframes together, save it in a new dataframe named 'd3'
d3 = pd.concat([d1, d2], ignore_index=True)
print(d3)
# Select the part of the dataframe such that it contains cities wherer temp is less then or equal to 20
print(d3[d3['Temperature'] <= 20])
# How many cities are there?
print(len(d3[d3['Temperature'] <= 20]))
# Select the part of the dataframe such that it contains the cities where tempearature greater than or equal to 35
print(d3[d3['Temperature'] >= 35])
# write function here
def is_ut(x):
    return x in ['Chandigarh', 'Pondicherry', 'Delhi']

d3['is_ut'] = d3['city'].apply(is_ut)
print(d3)
# write function here
def c_to_f(temp):
    return temp * 9/5 + 32
# check function c_to_f(10)
print(c_to_f(10))
# apply function c_to_f to d3 to create a column 'temp_farenhiet'
d3['temp_farenhiet'] = d3['Temperature'].apply(c_to_f)
print(d3)
# Select subset of the dataframe d1 such that it contains the cities which are union territories.
print(d1[d1['city'].isin(['Chandigarh', 'Delhi'])])
# Select a subset of the dataframe d1 such that it contains the cities which only have temperature above 90 Farenhiet.
print(d1[d1['Temperature'].apply(c_to_f) > 90])
# Select only the first three rows of the dataframe d1.
print(d1.head(3))
# Select all the rows and last two columns in the dataframe.
print(d3.iloc[:, -2:])
# Create a dataframe using dictionary of your choice
data = {
    'team': ['A', 'A', 'B', 'B', 'C', 'C'],
    'score': [10, 20, 30, 40, 50, 60],
    'player': ['P1', 'P2', 'P3', 'P4', 'P5', 'P6']
}
df2 = pd.DataFrame(data)
print(df2)
# Use Groupby of single column with aggregate sum()
print(df2.groupby('team')['score'].sum())
# Use Groupby of single column with aggregate count()
print(df2.groupby('team')['score'].count())
# Use Groupby of single column with aggregate min() and max()
print(df2.groupby('team')['score'].agg(['min', 'max']))
# Use Groupby of any 2 columns with aggregate mean()
print(df2.groupby(['team', 'player'])['score'].mean())
# Use Groupby of any 2 columns with aggregate min() and max()
print(df2.groupby(['team', 'player'])['score'].agg(['min', 'max']))
# Create a pandas daterange where starting date is 1st of January,2020 and end date is 1st of April 2021, store it in a new variable named 'a'
a = pd.date_range(start='2020-01-01', end='2021-04-01')
# print a
print(a)
# What is the len of a?
print(len(a))
# What is the type of a?
print(type(a))
```
